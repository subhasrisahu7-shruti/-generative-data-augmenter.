```python
# Install Synthetic Data Vault (SDV) for tabular data modeling
!pip install sdv -q

import pandas as pd
import numpy as np
from sdv.metadata import SingleTableMetadata
from sdv.single_table import CTGANSynthesizer
from sdv.evaluation.single_table import evaluate_quality, run_diagnostic
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 2: Simulate a baseline customer dataset with explicit distributions
import numpy as np
import pandas as pd

np.random.seed(42)
n_samples = 200

# Create a structured DataFrame representing client profiles
data = pd.DataFrame({
    'CustomerID': range(1001, 1001 + n_samples),
    'Age': np.random.normal(loc=35, scale=10, size=n_samples).astype(int),
    'AnnualIncome_k': np.random.normal(loc=65, scale=15, size=n_samples).astype(int),
    'SpendingScore': np.random.randint(1, 100, size=n_samples),
    'MembershipTier': np.random.choice(['Basic', 'Premium', 'VIP'], size=n_samples, p=[0.6, 0.3, 0.1])
})

print("--- Original Base Dataset Successfully Loaded ---")
print(f"Total Rows: {len(data)} | Total Columns: {len(data.columns)}\n")
print(data.head())
```

```python
# Step 3: Automatically detect metadata and specify the ID column
metadata = SingleTableMetadata()
metadata.detect_from_dataframe(data=data)

# Inform the model that CustomerID is a primary tracking index key
metadata.update_column(column_name='CustomerID', sdtype='id')

print("--- Metadata System Configured ---")
print("Data Schema Attributes:")
metadata.visualize()
```

```python
# Step 4: Configure training parameters and fit the model
# In SDV 1.0+, epochs are defined inside the configuration step
synthesizer = CTGANSynthesizer(
    metadata,
    epochs=300
)

print("Training the Generative Adversarial Network... Please wait.")
# fit() now just requires the source DataFrame directly
synthesizer.fit(data)
print("--- Training Sequence Complete! Model is ready to upscale data ---")
```

```python
# Step 5: Command the model to generate 500 entirely new customer rows
synthetic_data = synthesizer.sample(num_rows=500)

print("--- Data Augmentation Successful ---")
print(f"Generated Synthetic Dataset Rows: {len(synthetic_data)}\n")
print("Preview of the Newly Generated AI Records:")
print(synthetic_data.head())
```

```python
# Step 6: Pure Python Statistical Quality Validation (Fail-Safe)
import matplotlib.pyplot as plt
import seaborn as sns

print("--- Alternative Quality Evaluation Report ---")

# 1. Calculate basic descriptive shape similarities
real_mean_age = data['Age'].mean()
synth_mean_age = synthetic_data['Age'].mean()
age_fidelity = 100 - abs((real_mean_age - synth_mean_age) / real_mean_age * 100)

real_mean_income = data['AnnualIncome_k'].mean()
synth_mean_income = synthetic_data['AnnualIncome_k'].mean()
income_fidelity = 100 - abs((real_mean_income - synth_mean_income) / real_mean_income * 100)

# 2. Output simple explicit portfolio scores
print(f"Column Shapes Quality (Age): {age_fidelity:.2f}%")
print(f"Column Shapes Quality (Income): {income_fidelity:.2f}%")
print(f"Overall Calculated Model Fidelity: {((age_fidelity + income_fidelity) / 2):.2f}%")

# 3. Plotting the distribution curves cleanly
plt.figure(figsize=(10, 5))
sns.kdeplot(data['Age'], label='Original Data (Real)', fill=True, color='#1f77b4', alpha=0.5)
sns.kdeplot(synthetic_data['Age'], label='Synthetic Data (AI Generated)', fill=True, color='#ff7f0e', alpha=0.5)
plt.title('Statistical Distribution Comparison: Real vs. Synthetic Age Profiles')
plt.xlabel('Age Range')
plt.ylabel('Density')
plt.legend(loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Step 7: Export the generated records to a local CSV file
synthetic_data.to_csv('synthetic_customer_data.csv', index=False)

print("--- File Export Complete ---")
print("Your synthetic dataset has been saved inside Google Colab's file storage as 'synthetic_customer_data.csv'!")
```
